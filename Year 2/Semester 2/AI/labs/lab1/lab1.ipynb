```python
import random

def spinner(probabilityDistribution):
    print("Probabilities: " + str(probabilityDistribution))
    rand = random.random()
    print("Random number: " + str(rand))
    sum = 0

    for i, p in enumerate(probabilityDistribution):
        sum += p
        if rand < sum:
            return i


probabilities = [0.1, 0.7, 0.2]
spinner(probabilities)

```

Consider a large pipe with a radius R=30cm that has the opening pointing straight up, as in the figure. Around this pipe is a rectangular wall with an edge l=1m. Balls of radius r=1cm fall down uniformly from above inside the square yard. Compute, using a Monte Carlo simulation, the probability that a ball falls inside the pipe. Compare the results with the mathematically deduced probability.

SOLUTION:

Mathematically speaking:
A square = As = 1^2 = 1
Pipe radius = R = 0.3 m
Ball radius = r = 0.1 m

=> Ball must fall into a circle of the radius R-r = 0.29, and the A = Ac = pi(R-r)^2

=> probability = Ac/As = pi(R-r)^2 = 0.264


```python
l = 1.0
r = 0.01
R = 0.30

def simulation(N):
    got_in = 0
    for i in range(N):
        #origin at 0,0 so square coords are +-l/2
        x = random.uniform(-0.5, 0.5)
        y = random.uniform(-0.5, 0.5)

        # if distance from origin to coords <= radius of the circle, we good
        if x**2 + y**2 <= (R-r)**2:
            got_in += 1

    print("Result: " + str(got_in/N))

# nb of simulations
N = 1000000
simulation(N)
```
